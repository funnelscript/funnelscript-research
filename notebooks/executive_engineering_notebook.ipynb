## 🧠 1. The Reality Gap & The Bayesian Link

### Ingesting PyMC Marketing Posterior Distributions
Rather than hardcoding arbitrary human traits, virtual agents are initialized by sampling orthogonal parameter vectors from Bayesian posterior distributions derived from historical company data.

````python
import numpy as np

# Set deterministic random seed for bit-for-bit replayability
np.random.seed(42)
N_AGENTS = 100_000

# Bayesian Posterior Distributions from PyMC Marketing MCP
# 1. Baseline conversion propensity alpha ~ Beta(2, 8)
alpha = np.random.beta(2.0, 8.0, size=N_AGENTS)

# 2. Price sensitivity beta ~ Beta(3, 5)
beta = np.random.beta(3.0, 5.0, size=N_AGENTS)

# 3. FOMO index gamma ~ Beta(1.5, 4.0)
gamma = np.random.beta(1.5, 4.0, size=N_AGENTS)

# 4. Liquid category wallet balance (USD) ~ LogNormal(mu=4.5, sigma=0.6)
wallets = np.random.lognormal(mean=4.5, sigma=0.6, size=N_AGENTS)

# Assemble into contiguous 2D NumPy Memory Tensor
# Shape: (N_AGENTS, 4) -> [alpha, beta, gamma, wallet]
agent_tensor = np.column_stack([alpha, beta, gamma, wallets])
print(f"Contiguous Agent Tensor Initialized: {agent_tensor.shape}")
print(f"Mean Baseline Alpha: {alpha.mean():.4f}, Mean Wallet: ${wallets.mean():.2f}")
````

## 🛠️ 2. Subsystem Demonstration: Non-Linear Hill Saturation & Ad Fatigue

Marketing channels exhibit saturating marginal returns governed by the Hill equation:
$$\text{Saturation}(x) = \frac{x^n}{K^n + x^n}$$
where $K$ is the half-saturation point and $n$ is the Hill slope exponent.

````python
def hill_saturation(spend: np.ndarray, K: float = 10_000.0, slope: float = 2.0) -> np.ndarray:
    """Compute non-linear channel saturation efficiency in [0, 1]."""
    spend_pow = np.power(spend, slope)
    k_pow = np.power(K, slope)
    return spend_pow / (k_pow + spend_pow)

def ad_fatigue_multiplier(exposures: np.ndarray, decay_rate: float = 0.25) -> np.ndarray:
    """Compute exponential ad creative fatigue multiplier in [0, 1]."""
    return np.exp(-decay_rate * exposures)

# Demonstration across spend range
spend_curve = np.linspace(0, 50_000, 100)
yield_curve = hill_saturation(spend_curve, K=15_000.0, slope=2.0)
print(f"At $5,000 spend: Yield = {yield_curve[10]:.4f}")
print(f"At $15,000 spend (K): Yield = {yield_curve[30]:.4f}")
print(f"At $45,000 spend: Yield = {yield_curve[90]:.4f}")
````

## ⚡ 3. High-Throughput Vectorized Decision Kernel

Rather than executing slow object-oriented loops over $100,000$ agents, the transition kernel evaluates all agents simultaneously using parallel NumPy SIMD broadcasting.

````python
PRODUCT_PRICE = 99.00
DISCOUNT_RATE = 0.15
EFFECTIVE_PRICE = PRODUCT_PRICE * (1.0 - DISCOUNT_RATE)

# Extraction from contiguous tensor
a_alpha = agent_tensor[:, 0]
a_beta = agent_tensor[:, 1]
a_gamma = agent_tensor[:, 2]
a_wallet = agent_tensor[:, 3]

# Vectorized Purchase Probability Kernel
# P(buy) = alpha * (1 - beta * (Price / Wallet)) + gamma * Discount
relative_price = np.clip(EFFECTIVE_PRICE / np.maximum(a_wallet, 1.0), 0.0, 1.0)
p_purchase = a_alpha * (1.0 - 0.5 * a_beta * relative_price) + 0.3 * a_gamma * DISCOUNT_RATE
p_purchase = np.clip(p_purchase, 0.0, 1.0)

# Financial Invariant (Wallet Bound): Cannot purchase if price > liquid wallet
solvent_mask = (a_wallet >= EFFECTIVE_PRICE)
p_purchase = np.where(solvent_mask, p_purchase, 0.0)

# Stochastic Monte Carlo Transition Gate via Uniform Draw
u_draws = np.random.uniform(0.0, 1.0, size=N_AGENTS)
purchased_mask = (u_draws < p_purchase)

total_purchases = np.sum(purchased_mask)
total_revenue = total_purchases * EFFECTIVE_PRICE
print(f"Total Agents Evaluated: {N_AGENTS:,}")
print(f"Total Purchases: {total_purchases:,} (Conversion Rate: {total_purchases / N_AGENTS:.2%})")
print(f"Total Campaign Revenue: ${total_revenue:,.2f}")
````

## 🛡️ 4. Invariant Verification & Mass Balance

Every FunnelScript simulation strictly guarantees:
1. **Audience Mass Conservation:** $\sum_{s \in S} N_s(t) = N_{\text{total}}$
2. **Liquid Reserve Survival:** Value-at-Risk ($\text{VaR}_{95}$) calculation to identify cash starvation cliffs.

````python
# Verify Audience Mass Conservation
n_converted = np.sum(purchased_mask)
n_unconverted = np.sum(~purchased_mask)
assert (n_converted + n_unconverted) == N_AGENTS, "Audience Mass Balance Violated!"
print(f"✓ Invariant Verified: N_converted ({n_converted:,}) + N_unconverted ({n_unconverted:,}) == N_total ({N_AGENTS:,})")

# Risk Telemetry: Simulate 1,000 Monte Carlo Portfolio Draws
trial_conversions = np.random.binomial(n=N_AGENTS, p=p_purchase.mean(), size=1000)
trial_revenues = trial_conversions * EFFECTIVE_PRICE
var_95 = np.percentile(trial_revenues, 5)
cvar_95 = trial_revenues[trial_revenues <= var_95].mean()

print(f"Mean Revenue: ${trial_revenues.mean():,.2f}")
print(f"5th Percentile (VaR 95%): ${var_95:,.2f}")
print(f"Expected Shortfall (CVaR 95%): ${cvar_95:,.2f}")
````
